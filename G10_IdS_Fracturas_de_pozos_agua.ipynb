### Adquisición, construcción y calidad

**Fuente principal.** Se descarga manualmente el CSV de actualización diaria del Adjunto IV y se conserva sin editar. Cada ejecución registra nombre, tamaño, SHA-256 y fecha máxima de `fecha_data`; así se puede identificar exactamente la versión utilizada.

**Alternativa construida.** Como extensión, se puede unir el Adjunto IV con la producción no convencional mediante `idpozo` y agregar la producción de los primeros seis meses posteriores a `fecha_fin_fractura`. Requiere resolver pozos con múltiples intervenciones, meses faltantes y ventanas incompletas, por lo que no forma parte del modelo principal.

**Desafíos de calidad.** Se controlarán duplicados de `id_base_fractura_adjiv`, valores no numéricos, ceros, fechas inválidas o invertidas, categorías poco frecuentes, pozos repetidos y valores extremos. Los filtros se aplican de manera secuencial para que el resumen de exclusiones sea reproducible. No se imputará la variable objetivo.

```python
import hashlib
from pathlib import Path
from urllib.request import urlretrieve

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from IPython.display import display  # ty: ignore[unresolved-import]
```

```python
# @title Obtener el dataset en Colab/repositorio

dataset_url = "http://datos.energia.gob.ar/dataset/71fa2e84-0316-4a1b-af68-7f35e41f58d7/resource/2280ad92-6ed3-403e-a095-50139863ab0d/download/datos-de-fractura-de-pozos-de-hidrocarburos-adjunto-iv-actualizacin-diaria.csv"

NOMBRE_ARCHIVO = "datos-de-fractura-de-pozos-de-hidrocarburos-adjunto-iv-actualizacin-diaria.csv"

dataset_repo_path = Path("data") / NOMBRE_ARCHIVO
dataset_colab_path = Path("/content/data") / NOMBRE_ARCHIVO

if dataset_repo_path.exists():
	DATA_PATH = dataset_repo_path
elif dataset_colab_path.exists():
	DATA_PATH = dataset_colab_path
else:
	dataset_repo_path.parent.mkdir(parents=True, exist_ok=True)
	urlretrieve(dataset_url, dataset_repo_path)
	DATA_PATH = dataset_repo_path

print(f"Dataset disponible en: {DATA_PATH}")
```

```python
# @title Carga de dataframe

df_raw = pd.read_csv(DATA_PATH)
```

```python
sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", 50)


sha256 = hashlib.sha256(DATA_PATH.read_bytes()).hexdigest()

columnas_numericas = [
	"longitud_rama_horizontal_m",
	"cantidad_fracturas",
	"agua_inyectada_m3",
	"arena_bombeada_nacional_tn",
	"arena_bombeada_importada_tn",
	"co2_inyectado_m3",
	"presion_maxima_psi",
	"potencia_equipos_fractura_hp",
]
columnas_fecha = ["fecha_inicio_fractura", "fecha_fin_fractura", "fecha_data"]

df = df_raw.copy()
for col in columnas_numericas:
	df[col] = pd.to_numeric(df[col], errors="coerce")
for col in columnas_fecha:
	df[col] = pd.to_datetime(df[col], errors="coerce")

pasos = []


def aplicar_filtro(nombre, condicion):
	global df
	antes = len(df)
	df = df.loc[condicion].copy()
	pasos.append({"paso": nombre, "conservados": len(df), "excluidos_en_paso": antes - len(df)})


pasos.append({"paso": "Archivo original", "conservados": len(df), "excluidos_en_paso": 0})
aplicar_filtro("Cuenca Neuquina", df["cuenca"].eq("NEUQUINA"))
aplicar_filtro("Reservorio no convencional", df["tipo_reservorio"].eq("NO CONVENCIONAL"))
aplicar_filtro("Rama horizontal > 0", df["longitud_rama_horizontal_m"].gt(0))
aplicar_filtro("Cantidad de fracturas > 0", df["cantidad_fracturas"].gt(0))
aplicar_filtro("Agua inyectada > 0", df["agua_inyectada_m3"].gt(0))
aplicar_filtro("Fechas de operación válidas", df[["fecha_inicio_fractura", "fecha_fin_fractura"]].notna().all(axis=1))
aplicar_filtro("Fin no anterior al inicio", df["fecha_fin_fractura"].ge(df["fecha_inicio_fractura"]))

duplicados_id = int(df["id_base_fractura_adjiv"].duplicated().sum())
if duplicados_id:
	raise ValueError(f"Hay {duplicados_id} identificadores de fractura duplicados; deben revisarse.")

perfil_fuente = pd.DataFrame(
	{
		"indicador": ["archivo", "tamaño_bytes", "sha256", "filas_originales", "columnas", "fecha_data_máxima", "pozos_en_muestra"],
		"valor": [
			DATA_PATH.name,
			DATA_PATH.stat().st_size,
			sha256,
			len(df_raw),
			df_raw.shape[1],
			df_raw["fecha_data"].pipe(pd.to_datetime, errors="coerce").max(),
			df["idpozo"].nunique(),
		],
	}
)
display(perfil_fuente)
display(pd.DataFrame(pasos))
print(f"Conjunto analítico: {len(df):,} registros de {df['idpozo'].nunique():,} pozos.")
```

### Interpretación de la adquisición y la calidad

El archivo analizado contiene **4.910 registros y 30 variables**. La huella SHA-256 permite reconocer esta versión exacta del CSV, cuya fecha máxima de carga es el **29/09/2026**. Después de aplicar todos los criterios quedan **2.626 registros de 2.592 pozos**. La diferencia entre registros y pozos confirma que algunos pozos tienen más de una intervención; por eso `idpozo` se conserva como control y no como variable predictora.

Los filtros son secuenciales: `excluidos_en_paso` se interpreta respecto del paso anterior, no del archivo completo. Primero se retiraron 803 registros de otras cuencas y luego 376 correspondientes a reservorios convencionales dentro de la cuenca Neuquina. La mayor reducción posterior proviene de **1.071 registros sin una rama horizontal positiva**, coherente con el alcance limitado a pozos horizontales. También se excluyeron dos registros sin agua positiva y 32 con fecha de finalización anterior a la fecha de inicio.

El conjunto final representa aproximadamente el **53,5 %** del archivo original. Esta reducción no implica por sí sola mala calidad: una parte importante queda afuera por definición del alcance. En cambio, las fechas invertidas sí constituyen un problema de consistencia. Los resultados sólo describen el subconjunto retenido y no deben generalizarse a pozos convencionales, verticales u otras cuencas.

### Variable objetivo y algoritmo

La variable objetivo es `agua_inyectada_m3`. Se elige regresión porque el volumen conserva más información y responde mejor a la necesidad logística que una etiqueta arbitraria de “alto/bajo”.

`RandomForestRegressor` combina muchos árboles y puede capturar relaciones no lineales e interacciones sin exigir una ecuación previa. Es adecuado para integrar variables numéricas y categóricas y representar distintos perfiles de operación. Se comparará contra `DummyRegressor(strategy="median")`: el modelo agrega valor si reduce el error de esa referencia sencilla. La prueba temporal 2025–2026 se mantiene intacta durante el entrenamiento.

```python
from sklearn.compose import ColumnTransformer
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder

objetivo = "agua_inyectada_m3"
entradas_numericas = ["longitud_rama_horizontal_m", "cantidad_fracturas"]
entradas_categoricas = ["formacion_productiva", "subtipo_reservorio", "tipo_terminacion"]
entradas = entradas_numericas + entradas_categoricas

corte_temporal = pd.Timestamp("2025-01-01")
datos_modelo = df[entradas + [objetivo, "idpozo", "fecha_inicio_fractura", "id_base_fractura_adjiv"]].copy()
train = datos_modelo.loc[datos_modelo["fecha_inicio_fractura"] < corte_temporal].copy()
test = datos_modelo.loc[datos_modelo["fecha_inicio_fractura"] >= corte_temporal].copy()

pozos_compartidos = set(train["idpozo"]).intersection(test["idpozo"])
train = train.loc[~train["idpozo"].isin(pozos_compartidos)].copy()
if train.empty or test.empty:
	raise ValueError("La partición temporal quedó vacía; revisar cobertura de fechas.")

X_train, y_train = train[entradas], train[objetivo]
X_test, y_test = test[entradas], test[objetivo]

try:
	codificador = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
except TypeError:  # compatibilidad con versiones anteriores de scikit-learn
	codificador = OneHotEncoder(handle_unknown="ignore", sparse=False)

prep = ColumnTransformer(
	[
		("num", SimpleImputer(strategy="median"), entradas_numericas),
		(
			"cat",
			Pipeline(
				[
					("imputer", SimpleImputer(strategy="most_frequent")),
					("onehot", codificador),
				]
			),
			entradas_categoricas,
		),
	]
)

modelo = Pipeline(
	[
		("prep", prep),
		("regresor", RandomForestRegressor(n_estimators=400, min_samples_leaf=3, random_state=42, n_jobs=-1)),
	]
)
referencia = DummyRegressor(strategy="median")

modelo.fit(X_train, y_train)
referencia.fit(X_train, y_train)
pred_modelo = modelo.predict(X_test)
pred_referencia = referencia.predict(X_test)

print(f"Entrenamiento: {len(train):,} registros hasta 2024")
print(f"Prueba temporal: {len(test):,} registros de 2025–2026")
print(f"Pozos retirados del entrenamiento por aparecer en prueba: {len(pozos_compartidos)}")
```

### Interpretación de la partición

El entrenamiento utiliza **1.999 operaciones anteriores a 2025** y la prueba conserva **623 operaciones de 2025–2026**. Se retiraron del entrenamiento cuatro pozos que reaparecen en prueba. Esta separación es más exigente y más cercana a un uso futuro que una división aleatoria, porque el modelo aprende del pasado y se evalúa en un período posterior. También permite detectar cambios de tecnología o de prácticas operativas que una partición aleatoria podría ocultar.

### Métricas de evaluación

| Métrica | Interpretación |
|---|---|
| MAE | Error absoluto promedio en m³. Es la métrica principal por su lectura directa. |
| RMSE | Error en m³ que penaliza con mayor fuerza los desaciertos grandes. |
| Error mediano absoluto | Error típico menos sensible a valores extremos. |
| R² | Proporción de variación explicada respecto de predecir la media; puede ser negativa en prueba. |

La matriz de confusión no corresponde porque el problema final es de regresión. Además de las métricas, se revisarán errores por año y gráficos de observado frente a estimado.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, median_absolute_error, r2_score


def calcular_metricas(nombre, real, pred):
	return {
		"modelo": nombre,
		"MAE_m3": mean_absolute_error(real, pred),
		"RMSE_m3": np.sqrt(mean_squared_error(real, pred)),
		"MedAE_m3": median_absolute_error(real, pred),
		"R2": r2_score(real, pred),
	}


metricas = pd.DataFrame(
	[
		calcular_metricas("Referencia: mediana", y_test, pred_referencia),
		calcular_metricas("Random Forest", y_test, pred_modelo),
	]
)
display(metricas.round(3))

mae_ref = metricas.loc[metricas["modelo"].eq("Referencia: mediana"), "MAE_m3"].iloc[0]
mae_modelo = metricas.loc[metricas["modelo"].eq("Random Forest"), "MAE_m3"].iloc[0]
mejora = 100 * (mae_ref - mae_modelo) / mae_ref
decision = "mejora la referencia" if mejora > 0 else "no mejora la referencia"
print(f"Decisión preliminar: el Random Forest {decision} en MAE ({mejora:.1f}%).")

resultados_test = test[["id_base_fractura_adjiv", "idpozo", "fecha_inicio_fractura"]].copy()
resultados_test["agua_observada_m3"] = y_test.to_numpy()
resultados_test["agua_estimada_m3"] = pred_modelo
resultados_test["error_m3"] = resultados_test["agua_estimada_m3"] - resultados_test["agua_observada_m3"]
resultados_test["error_absoluto_m3"] = resultados_test["error_m3"].abs()
display(resultados_test.nlargest(10, "error_absoluto_m3").round(1))

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].scatter(y_test, pred_modelo, alpha=0.5, s=25)
limite = max(float(y_test.max()), float(pred_modelo.max()))
axes[0].plot([0, limite], [0, limite], "--", color="black", linewidth=1)
axes[0].set(title="Observado frente a estimado", xlabel="Agua observada (m³)", ylabel="Agua estimada (m³)")

sns.histplot(resultados_test["error_m3"], bins=35, ax=axes[1])
axes[1].axvline(0, color="black", linestyle="--", linewidth=1)
axes[1].set(title="Distribución del error", xlabel="Error: estimado − observado (m³)", ylabel="Registros")
plt.tight_layout()
plt.show()

nombres = modelo.named_steps["prep"].get_feature_names_out()
importancias = pd.Series(modelo.named_steps["regresor"].feature_importances_, index=nombres, name="importancia").sort_values(ascending=False)
display(importancias.head(12).to_frame().round(3))
print("Las importancias describen el uso de variables por el modelo; no prueban causalidad.")
```

### Interpretación de la evaluación

El Random Forest obtiene un **MAE de 9.758 m³**, frente a **29.428 m³** de la referencia basada en la mediana: reduce el error absoluto promedio en **66,8 %**. El error mediano absoluto baja de 26.767 a **6.532 m³**, por lo que en al menos la mitad de los casos el desacierto es relativamente acotado. El **R² de 0,469** indica que el modelo explica aproximadamente el 46,9 % de la variabilidad observada en 2025–2026; queda una parte importante sin explicar. El R² negativo de la referencia muestra que la mediana histórica se adapta mal al período de prueba.

El RMSE del modelo (**14.812 m³**) es bastante mayor que su MAE. Esto señala que existen unos pocos errores grandes que reciben mayor penalización. La tabla permite identificar esos casos: el mayor se sobreestima en unos **114.180 m³**, y aparecen tanto sobreestimaciones como subestimaciones superiores a 49.000 m³. Estos registros forman una lista concreta de operaciones para revisar y ayudan a definir cuándo la aplicación debe acompañar la estimación con antecedentes comparables.

**Gráfico observado frente a estimado.** La línea punteada representa una predicción perfecta. La mayoría de los puntos sigue una tendencia cercana a esa diagonal, especialmente en el rango medio. Sin embargo, la dispersión aumenta para volúmenes altos. También aparece una franja horizontal cercana a 118.000 m³, típica de árboles que asignan la misma estimación a varios casos y que no extrapolan con facilidad fuera de los patrones aprendidos.

**Gráfico de errores.** La distribución se concentra alrededor de cero, sin un sesgo general extremo. Aun así, presenta colas hacia ambos lados y una cola positiva más larga debido a algunas sobreestimaciones muy grandes. Un error positivo significa que el modelo estimó más agua que la observada; uno negativo significa que subestimó.
