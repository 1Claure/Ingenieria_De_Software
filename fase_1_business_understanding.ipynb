## Estudio de caso

Proyecto del Grupo 10 — versión 1.1 · 14/10/2026

***Tema de trabajo.***

El proyecto busca analizar y estimar el volumen de agua registrado en operaciones de fractura hidráulica de pozos horizontales no convencionales de la cuenca Neuquina. Para ello, se utilizarán datos públicos del Adjunto IV de la Secretaría de Energía.
La fuente reúne información sobre características de los pozos y sus operaciones de fractura. Estas variables permiten explorar diferencias entre operaciones y desarrollar un modelo de regresión que estime el volumen de agua registrado. La solución se plantea como una herramienta de apoyo para la planificación logística y la comparación con operaciones históricas; no reemplaza la evaluación técnica de los especialistas.

Dataset: [Datos de fractura de pozos — Adjunto IV.](https://datos.gob.ar/dataset/datos-de-fractura-de-pozos-adjunto-iv)


**Problema de negocio** El volumen de agua utilizado en una operación de fractura es relevante para planificar el abastecimiento, el transporte y la capacidad de almacenamiento. Los registros históricos permiten estudiar la asociación entre el volumen de agua y variables como la longitud de la rama horizontal, la cantidad de etapas y las características del reservorio y de la terminación.
El proyecto evaluará si estas variables permiten producir una estimación suficientemente precisa para ofrecer una referencia inicial de planificación. Las asociaciones observadas no se interpretarán automáticamente como relaciones causales.


> **Pregunta del estudio:** ¿con qué error puede estimarse el volumen de agua inyectada a partir de la longitud horizontal, la cantidad de etapas y las características geológicas y de terminación registradas?


**Objetivo general.** Desarrollar, siguiendo CRISP-DM, un modelo que estime `agua_inyectada_m3` en fracturas horizontales no convencionales de la cuenca Neuquina y comunicar su error y sus límites mediante un notebook y una aplicación web.

**Objetivos específicos**

1. Versionar el CSV oficial y auditar su estructura, cobertura y calidad.
2. Construir un conjunto reproducible con criterios explícitos de inclusión y exclusión.
3. Describir la relación entre agua, longitud horizontal, cantidad de fracturas, formación y terminación.
4. Comparar un `RandomForestRegressor` con una referencia que siempre predice la mediana.
5. Evaluar ambos modelos sobre operaciones de 2025–2026 mediante MAE, RMSE, error mediano absoluto y R².
6. Publicar un dashboard consistente con el notebook y documentar su actualización.

**Alcance y decisiones metodológicas**

| Aspecto | Decisión |
|---|---|
| Población | `cuenca=NEUQUINA`, `tipo_reservorio=NO CONVENCIONAL`, rama horizontal, etapas y agua mayores que cero. La cuenca Neuquina no equivale a la provincia de Neuquén. |
| Unidad de análisis | Un registro de fractura (`id_base_fractura_adjiv`). Se controlarán identificadores duplicados y pozos con más de un registro. |
| Variable objetivo | `agua_inyectada_m3`, cuantitativa continua. La clasificación de 50.000 m³ se conserva sólo como corte descriptivo, porque no es un estándar industrial y perdería información. |
| Entradas | `longitud_rama_horizontal_m`, `cantidad_fracturas`, `formacion_productiva`, `subtipo_reservorio` y `tipo_terminacion`. Son variables plausibles para un análisis de diseño, pero la fuente no distingue valores planificados de ejecutados. |
| Variables excluidas | Arena, CO₂, presión, potencia, identificadores, empresas y fechas. Pueden estar asociadas con el agua, pero no se garantiza que se conozcan antes de la operación o podrían introducir fuga de información. |
| Validación | Entrenar con operaciones anteriores a 2025 y reservar 2025–2026 como prueba temporal. Si un pozo aparece en ambos períodos, se elimina del entrenamiento para evitar compartir pozos. |
| Modelo | Mediana de entrenamiento como referencia y Random Forest como candidato. Si el modelo no reduce el MAE de la referencia, no se recomienda su uso. |
| Software | Streamlit con fuente y versión, calidad, exploración, estimación, error del conjunto reservado y rangos de referencia. |

**Visión general del análisis**

El proyecto parte de un registro público de operaciones de fractura hidráulica. El CSV contiene información del pozo, la cuenca, el yacimiento, la formación productiva, el tipo de reservorio, la longitud horizontal, la cantidad de etapas, el tipo de terminación, los volúmenes de agua, arena y CO₂, la presión, la potencia y las fechas de la operación. Cada fila representa una fractura informada ante la Secretaría de Energía.

Para responder una pregunta concreta de negocio, el análisis se concentra en fracturas horizontales no convencionales de la cuenca Neuquina. A partir de ese universo se busca estimar `agua_inyectada_m3`, una magnitud central para organizar abastecimiento, transporte, almacenamiento y coordinación de recursos.

El recorrido del proyecto es el siguiente:

1. **Comprender y depurar los registros:** identificar la versión del archivo, seleccionar el alcance y controlar fechas y valores.
2. **Explorar los antecedentes:** describir volúmenes típicos y estudiar su relación con la longitud lateral, la cantidad de fracturas, la formación y la terminación.
3. **Aprender un patrón histórico:** entrenar un modelo con operaciones anteriores a 2025.
4. **Comprobar su desempeño:** estimar operaciones de 2025–2026 que el modelo no utilizó para aprender.
5. **Llevarlo a una aplicación:** permitir que el usuario cargue las características de una operación, obtenga un volumen estimado y lo compare con antecedentes similares.

El producto final se plantea como una herramienta de apoyo a la planificación: transforma datos históricos dispersos en una referencia cuantitativa, reproducible y fácil de consultar.
